### 1.3 Exceptions

Write `safe_divide(a, b)` that:
- Returns `a / b` normally
- Returns `None` if `b == 0` (catch `ZeroDivisionError`)
- Raises `TypeError` with message `"Inputs must be numeric"` if either input is not a number

````python
def safe_divide(a, b):
    if not isinstance(a, (int,float)) or not isinstance(b,(int,float)):
        raise TypeError("Inputs must be numeric")
    try:
        return a/b
    except ZeroDivisionError:
        return None
    

# Tests
assert safe_divide(10, 2) == 5.0
assert safe_divide(5, 0) is None
try:
    safe_divide('x', 2)
    assert False, "Should have raised TypeError"
except TypeError as e:
    assert str(e) == "Inputs must be numeric"
print("1.3 passed ✓")
````

### 4.3 Eigenvalues & Eigenvectors

For the matrix:
```

     [2, 3]]
```

1. Compute eigenvalues and eigenvectors using `np.linalg.eig`.
2. **Verify** the eigenvector equation: `A @ v = λ * v` for each eigenpair.
3. Plot both eigenvectors as arrows, scaled by their eigenvalues, on a 2D plot.
4. **Explain** in a markdown cell (below): what does it geometrically mean for a matrix to stretch a vector?

````python
A = np.array([[4, 1], [2, 3]], dtype=float)

eigenvalues, eigenvectors = np.linalg.eig(A)

print(f"Eigenvalues: {eigenvalues}")
print(f"Eigenvectors (columns):\n{eigenvectors}")

# Verify Av = λv for each eigenpair
for i in range(len(eigenvalues)):
    lamda = eigenvalues[i]
    vec = eigenvectors[:, i] 
    Av  = A @ vec
    lv  = lamda * vec
    print(f"  λ={lamda:.2f}: A@v = {Av.round(6)},  λ*v = {lv.round(6)},  match={np.allclose(Av, lv)}")

# Plot
fig, ax = plt.subplots(figsize=(5, 5))
colors = ["blue", "red"]
for i in range(2):
    lamda = eigenvalues[i]
    vec = eigenvectors[:, i]
    ax.quiver(0, 0, vec[0]*lamda, vec[1]*lamda, angles='xy', scale_units='xy', scale=1,
              color=colors[i], width=0.01, label=f'λ={lamda:.1f}')
    ax.quiver(0, 0, vec[0], vec[1], angles='xy', scale_units='xy', scale=1,
              color="black", width=0.01, alpha=0.7, label=f'eigenvector {i+1}')
ax.set_xlim(-4, 4); ax.set_ylim(-4, 4)
ax.axhline(0, color='k', lw=0.6); ax.axvline(0, color='k', lw=0.6)
ax.set_title('Eigenvectors scaled by eigenvalues')
ax.set_xlabel('x'); ax.set_ylabel('y')
ax.legend(); plt.tight_layout(); plt.show()    

print("4.3 passed ✓")
````

> 
Geometrically, a matrix transformation can be imagined as something that stretches, shrinks or rotates vectors in space. What matrix does is, it acts as a transformation machine. It takes a vector and transforms it into another vector. Equation for matrix transformation is:

       Av = w

    Here:
        -A is the matrix
        -v is old vector
        -w is new vector 


When a transformation happens both direction and size of vector changes.


Here comes the concept of eigen values and eigen vectors. 


Eigen Vectors are special type of vectors that keeps its original direction after transformation and only its size changes.


Eigen Values is the change in size of the vector i.e. it tells how much stretching or squishing happens.


Equation for Eigenvalues and Eigenvectors is:


        Av = λv


    Here:


        -A is the matrix
        -v is eigen vector(Direction remains unchanged)
        -λ is eigen value  

 
 Geometric meaning of Stretching:


    Stretching = increase in size


 For example say we have a vector with length 2 and λ = 3 then vector becomes λ*v i.e. three times of the original size.


  Geometric meaning of Squishing:  


     Squishing = decrease in size


 For example say we have a vector with length 2 and λ = 0.5 then vector becomes λ*v i.e. half of the original size.
